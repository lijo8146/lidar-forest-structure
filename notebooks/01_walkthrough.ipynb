# LiDAR forest structure: inspect an actual run
Run CLI processing first. This uses real local products and fails if missing. Synthetic test fixtures are separate.

```python
from pathlib import Path
from IPython.display import Image, display
from lidarforest.workflow import validate
root = Path.cwd()
if not (root / 'config/example.yml').exists():
    root = root.parent
out = root / 'data/processed'
if not (out / 'run.json').exists():
    raise FileNotFoundError('Run lidar-forest download and lidar-forest run first.')
run = validate(out)
print(f"Validated {run['aoi_points']:,} AOI returns")
```

## Two different foot units
Inspect horizontal/vertical metadata separately. NAVD88 is retained; no geoid transformation is applied.

```python
coordinates = run['ingestion']['coordinates']
for key in ['horizontal_operation', 'vertical_to_metre', 'vertical_operation']:
    print(key, ':', coordinates[key])
```
